```python
import json
```

```python
def load_data(file_name):
    with open(file_name,"r") as f:
        data=json.load(f)
    
    return data
```

```python
data=load_data("store_data.json")
```

```python
# Asssignment number 1.
```

```python
# Clean and structure data
def clean_data(data):
    text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5}
    clean_datas=[];
    unique_user=set();
    #clean rating
    for user in data:
        raw_rating = user["rating"] #.strip().lower()

        if(raw_rating in text_to_num):
            raw_rating=text_to_num[raw_rating]
        user["rating"]=raw_rating;

        #handle missing value

        raw_age=user.get("age")
        if(raw_age==None):
            user["age"]=None;

        #Duplication

        if(user["name"].strip() in unique_user):
            continue;
        unique_user.add(user["name"].strip())
        clean_datas.append(user);


    return clean_datas;
        
        
        

        
```

```python
new_data=clean_data(data);
```

```python
print(new_data);
```

```python
# Asssignment number 2.

def average_rating(new_data):
    rating_count=0;
    count=0;
    for user in new_data:
        rating_count+=float(user["rating"]);
        count+=1;

    print(rating_count)
    print(count)

    average=rating_count/count;
    print(average);

#what % user give poor rating less than 3

def poor_rating_per(new_data):
    poor_rating_count=0;
    count=0;
    for user in new_data:
        rating=float(user["rating"]);
        if(rating <3):
            poor_rating_count+=1;
        count+=1;
    
    poor_rating_percentage=(poor_rating_count*100)/count;
    print(poor_rating_percentage)
    print( f"{poor_rating_percentage} %")
        
        
```

```python
average_rating(new_data)
poor_rating_per(new_data)
```

```python
#Build product recommendation feature(rule base not ml base);

def recommendation_feature(new_data):
    recommendation=[];

    for user in new_data:
        curr_rec={};
        curr_rec["name"]=user["name"];
        rating=float(user["rating"]);
        if(rating >=4):
            curr_rec["brand"]="Apple";
        else:
            curr_rec["brand"]="Summsung";

        recommendation.append(curr_rec);

    return recommendation;
            
    
```

```python
data=recommendation_feature(new_data);

for i in data:
    print(i)
```
