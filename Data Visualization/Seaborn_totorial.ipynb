```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
tips=sns.load_dataset("tips")
tips.head()
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    size="size",
    style="smoker"
)
```

```python
#scatter plot


tips=sns.load_dataset("tips")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",

    hue="time"
)
```

```python
#line plot
tips=sns.load_dataset("tips")

sns.lineplot(
    data=tips,
    x="total_bill",
    y="tip",

    hue="time"
)
```

```python
# Bar Plot

tips=sns.load_dataset("tips")
tips.head()

sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
# Box Plot

tips=sns.load_dataset("tips")
tips.head()

sns.boxplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
penguins=sns.load_dataset("penguins")
penguins.head()
sns.histplot(

    data=penguins,
    bins=20,
    x=("body_mass_g")
)
```

```python

flight=sns.load_dataset("flights")

flight.head()

flight_pivot=flight.pivot(index="month",columns="year" ,values="passengers")

sns.heatmap(
    flight_pivot,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)

plt.title("heatmap")
plt.tight_layout()
```

```python
fig, ax=plt.subplots();

sns.lineplot(
    data=tips,

    x="day",
    y="total_bill",
    hue="sex",
    ax=ax,
errorbar=None
)

ax.set_title("Tips analysis")
ax.set_xlabel("days"),
ax.set_ylabel("total bill")

fig.tight_layout()
    
```
