```python
# Import Matplotlib for data visualization
import matplotlib.pyplot as plt

# Import NumPy for numerical operations
import numpy as np

# Set random seed to generate the same random values every time
np.random.seed(0)

# Generate 1000 random values using Normal Distribution
# 70 = Mean (average)
# 10 = Standard Deviation (spread of data)
# 1000 = Number of random values
score = np.random.normal(70, 10, 1000)

# Create a histogram to visualize the distribution of scores
# edgecolor="black" adds black borders to each bar
plt.hist(score, edgecolor="black")

# Display the histogram
plt.show()
```
