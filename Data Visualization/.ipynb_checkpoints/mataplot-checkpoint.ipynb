```python
# Plot multiple
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_years = [2008, 2009, 2010, 2011, 2012]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot(year, oscar_revenue, "o--r" ,label="Oscar Winners")
plt.plot(non_oscar_years, non_oscar_revenue, label="Non-Oscar Winners")
plt.title("Years vs Revenue(in $M)")
plt.xlabel("Years")
plt.ylabel("Revenue(in $M)")
plt.legend()
# plt.legend(loc="upper right")
# plt.legend(loc="upper left")
# plt.legend(loc="center")
```

```python
import matplotlib.pyplot as plt

non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_year = [2008, 2009, 2010, 2011, 2012]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.bar(non_oscar_year, non_oscar_revenue, color="orange")

plt.title("Non-Oscar Year and Non-Oscar Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue ($M)")

for i in range(len(non_oscar_year)):
    plt.text(
        non_oscar_year[i],
        non_oscar_revenue[i] + 20,
        str(non_oscar_revenue[i]),
        ha="center"
    )

plt.ylim(0, max(non_oscar_revenue) + 100)

plt.show()
```
