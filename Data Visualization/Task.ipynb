```python
import matplotlib.pyplot as plt
import numpy as np

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temp = [
    [22, 23, 21, 24, 25],
    [18, 19, 17, 20, 21],
    [30, 32, 31, 33, 34],
    [25, 26, 24, 27, 28]
]

fig, axis = plt.subplots(2, 2, figsize=(10, 7))

city = 0

for i in range(2):
    for j in range(2):

        axis[i][j].plot(
            days,
            temp[city],
            marker="o",
            linestyle="--"
        )

        axis[i][j].set_title(cities[city])
        axis[i][j].set_xlabel("Days")
        axis[i][j].set_ylabel("Temperature (°C)")
        axis[i][j].grid(True)

        city += 1

fig.suptitle("Weekly Temperature Comparison", fontsize=16)

fig.tight_layout()
plt.show()
```
