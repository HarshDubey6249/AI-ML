```python
import numpy as np
import matplotlib.pyplot as plt

# Non-Oscar Winner Movies
non_oscar_movies = [
    "Slumdog Millionaire",
    "Avatar",
    "Inception",
    "Hugo",
    "Lincoln"
]

year = [2008, 2009, 2010, 2011, 2012]

non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M


# Oscar Winning Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

oscar_revenue = [1005, 170, 427, 133, 232]  # in $M


# X positions
x = np.arange(len(year))
width = 0.4


# Title and labels
plt.title("Oscar and Non-Oscar Movie Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue ($M)")


# Bar charts
plt.bar(
    x - width/2,
    oscar_revenue,
    width,
    color="green",
    label="Oscar Movies"
)

plt.bar(
    x + width/2,
    non_oscar_revenue,
    width,
    label="Non-Oscar Movies"
)


# Show years on X-axis
plt.xticks(x, year)

# Show legend
plt.legend()

# Display chart
plt.show()
```

```python

# Scatter Plot Data
people = [
    "Person A", "Person B", "Person C", "Person D", "Person E",
    "Person F", "Person G", "Person H", "Person I", "Person J"
]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]

blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]


# Create colors based on BP
colors = ["green" if bp < 135 else "red" for bp in blood_pressure]

# Create scatter plot
plt.scatter(
    age,
    blood_pressure,
    color=colors,
    s=blood_pressure,
    marker="^",
    alpha=0.5
)

# Title and labels
plt.title("Age vs Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

# Grid
plt.grid()

# Add person names
for i in range(len(age)):
    plt.annotate(
        people[i],
        xy=(age[i], blood_pressure[i])
    )

# Set X-axis range
plt.xlim(min(age), max(age) + 10)

# Display
plt.show()
```
