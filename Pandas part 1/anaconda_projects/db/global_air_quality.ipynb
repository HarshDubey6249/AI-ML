```python
import pandas as pd 
```

```python
df=pd.read_csv("globalAirQuality.csv");
print(pd.DataFrame(df))
```

```python
# ==============================
# Select Columns
# ==============================

df["country"]                  # Select a single column
df[["city", "aqi"]]            # Select multiple columns


# ==============================
# Select Rows by Label (.loc)
# ==============================

df.loc[0]                      # Select the row with label 0
df.loc[0:3]                    # Select rows from label 0 to 3 (both inclusive)


# ==============================
# Select Rows by Position (.iloc)
# ==============================

df.iloc[0]                     # Select the first row (position 0)
df.iloc[4:7]                   # Select rows at positions 4, 5, and 6
                                # End position 7 is exclusive


# ==============================
# Select Rows & Columns
# ==============================

df.loc[0, "country"]            # Select row with label 0 and the "country" column

df.iloc[0, 2]                   # Select first row and third column
                                # Position 0 = 1st row
                                # Position 2 = 3rd column


df.iloc[0, 1:5]                 # Select first row and columns at positions 1 to 4
                                # End position 5 is exclusive

df.loc[0, ["country", "aqi"]]   # Select row 0 and specific columns


# Select multiple rows and specific columns using .loc
df.loc[0:5, ["country", "city", "latitude"]]
# Rows 0 to 5 are selected (both inclusive)
# Only "country", "city", and "latitude" columns are selected


# Select multiple rows and columns using .iloc
df.iloc[0:5, 1:3]
# Rows at positions 0 to 4 are selected
# Columns at positions 1 to 2 are selected
# End positions are exclusive


# ==============================
# Select a Single Scalar Value
# ==============================

df.at[0, "city"]                # Select one value using row label and column name

df.iat[0, 2]                    # Select one value using row position and column position
```

```python
# =========================
# Filtering Data
# =========================

# Select rows where AQI is greater than 100
df[df["aqi"] > 100]

# Select rows where AQI is greater than 100
# and display only the "city" and "aqi" columns
df[df["aqi"] > 100][["city", "aqi"]]


# Filter rows where:
# 1. AQI is greater than 100
# 2. Timestamp matches the given date and time
#
# & is used for AND condition in Pandas
# Each condition must be written inside parentheses
df[
    (df["aqi"] > 100) &
    (df["timestamp"] == "2025-11-04 18:25:17.554219")
]


# Store the filtered data in a variable
# Only keep the "city" and "aqi" columns
aqi_data = df[
    (df["aqi"] > 100) &
    (df["timestamp"] == "2025-11-04 18:25:17.554219")
][["city", "aqi"]]


# =========================
# Difference between loc & iloc
# =========================

# .loc[] selects data using the ROW LABEL / INDEX LABEL
aqi_data.loc[360]

# .iloc[] selects data using the INTEGER POSITION
# 1 means the second row
aqi_data.iloc[1]

# Note:
# .loc[360] and .iloc[1] can return the same row/value
# ONLY if row label 360 happens to be at position 1.


# =========================
# Query
# =========================

# .query() is another way to filter DataFrame rows
# It returns a new DataFrame (copy-like result)
df.query(
    "aqi > 100 & timestamp == '2025-11-04 18:25:17.554219'"
)


# Store a value in a variable
my_country = "IN"

# @my_country allows us to use a Python variable inside query()
# This selects rows where country is equal to "IN"
df.query("country == @my_country")


# =========================
# Working with Timestamp
# =========================

# Convert the "timestamp" column from string/object
# into Pandas datetime format
df["timestamp"] = pd.to_datetime(df["timestamp"])


# Extract only the date from the timestamp
# Example:
# 2025-11-04 18:25:17
# becomes:
# 2025-11-04
df["date"] = df["timestamp"].dt.date


# Convert the selected date into a Python date object
selected_date = pd.to_datetime("2025-11-04").date()


# Filter rows where:
# 1. Country is equal to my_country
# 2. Date is equal to selected_date
# 3. AQI is greater than 100
df.query(
    "country == @my_country & date == @selected_date & aqi > 100"
)


# =========================
# Using .copy()
# =========================

# Select only "country" and "aqi" columns
# .copy() creates an independent DataFrame
#
# This helps avoid confusion between a view and a copy
countries = df[["country", "aqi"]].copy()


# Change the "country" value to "India"
# for rows with index 0 through 4
#
# .loc[] is used to select rows and columns by labels
countries.loc[0:4, "country"] = "India"


# Display the modified DataFrame
print(countries)

```

```python
# Cleaning and Handling Missing Values

# Read the CSV file and store the data in a DataFrame
df = pd.read_csv("raw_data.csv")


# Check where the missing (NaN) values are present
# Returns True where a value is missing and False where a value exists
df.isnull()


# Count the number of missing values in each column
df.isnull().sum()


# Remove rows that contain at least one missing value
# axis=0 → work with rows (default)
df.dropna()


# Remove columns that contain at least one missing value
# axis=1 → work with columns
df.dropna(axis=1)


# Replace all missing values with 0
# df.fillna(0)


# Replace missing values in the "age" column
# with the average (mean) age
# df_mean = df["age"].fillna(df["age"].mean())
# print(df_mean)


# Create a copy of the original DataFrame
# This allows us to clean the data without changing the original df
cleaned_data = df.copy()


# Fill missing values in the "age" column
# with the mean (average) age of the column
cleaned_data["age"] = cleaned_data["age"].fillna(
    cleaned_data["age"].mean()
)


# Display the cleaned DataFrame
cleaned_data


# Forward Fill (ffill)
# Fill missing values using the previous available value
df.ffill()


# Backward Fill (bfill)
# Fill missing values using the next available value
df.bfill()
```
