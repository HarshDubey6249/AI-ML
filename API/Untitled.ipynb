```python
import requests as rq
```

```python
URL="https://www.scrapethissite.com/pages/simple/"

res=rq.get(URL)
```

```python

if res.status_code == 200:
    print(res.headers)

    with open("scraprd_data/data1.html", "w", encoding="utf-8") as f:
        f.write(res.text)

    print("HTML file saved successfully!")
else:
    print("Request failed:", res.status_code)
```

```python
# Beautifulsoup

from bs4 import BeautifulSoup as bs

with open("scraprd_data/data1.html", "r", encoding="utf-8") as f:
     data= f.read()

soup=bs(data,"lxml");

```

```python
all_h3=soup.find_all("h3");

all_data=[];

for h3 in all_h3:
    name =h3.get_text(strip=True)

    population=h3.find_next("div").select("span.country-population")[0].get_text()
    

    all_data.append([name,population])
```

```python
import pandas as pd

df=pd.DataFrame(all_data,columns=["country","population"])
```
