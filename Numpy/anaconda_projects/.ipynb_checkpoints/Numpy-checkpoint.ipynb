```python
import numpy as np

# =====================================================
# NUMPY: DIFFERENT WAYS TO CREATE 2D ARRAYS (MATRICES)
# =====================================================

# 1. Create a 2D array manually
arr1 = np.array([[1, 2, 3],
                 [4, 5, 6]])
print("1. np.array()")
print(arr1)
print()


# 2. Create a matrix filled with zeros
arr2 = np.zeros((3, 4))      # 3 rows, 4 columns
print("2. np.zeros()")
print(arr2)
print()


# 3. Create a matrix filled with ones
arr3 = np.ones((2, 5))       # 2 rows, 5 columns
print("3. np.ones()")
print(arr3)
print()


# 4. Create an empty matrix (contains garbage values)
arr4 = np.empty((2, 3))
print("4. np.empty()")
print(arr4)
print()


# 5. Create an identity matrix
arr5 = np.eye(4)
print("5. np.eye()")
print(arr5)
print()


# 6. Create a diagonal matrix
arr6 = np.diag([10, 20, 30, 40])
print("6. np.diag()")
print(arr6)
print()


# 7. Create a matrix filled with a specific value
arr7 = np.full((3, 3), 7)
print("7. np.full()")
print(arr7)
print()


# 8. Create values using arange() and reshape()
arr8 = np.arange(12).reshape(3, 4)
print("8. np.arange() + reshape()")
print(arr8)
print()


# 9. Create equally spaced values using linspace()
arr9 = np.linspace(0, 1, 12).reshape(3, 4)
print("9. np.linspace() + reshape()")
print(arr9)
print()


# 10. Create random floating point numbers between 0 and 1
arr10 = np.random.rand(3, 3)
print("10. np.random.rand()")
print(arr10)
print()


# 11. Create random integers
arr11 = np.random.randint(1, 100, size=(3, 4))
print("11. np.random.randint()")
print(arr11)
print()


# 12. Create random numbers from normal distribution
arr12 = np.random.randn(3, 3)
print("12. np.random.randn()")
print(arr12)
print()


# 13. Create zeros with same shape as another array
arr13 = np.zeros_like(arr1)
print("13. np.zeros_like()")
print(arr13)
print()


# 14. Create ones with same shape as another array
arr14 = np.ones_like(arr1)
print("14. np.ones_like()")
print(arr14)
print()


# 15. Create a matrix with same shape and custom value
arr15 = np.full_like(arr1, 9)
print("15. np.full_like()")
print(arr15)
print()
```

```python
# vectorization

arr = np.array([1, 2, 3, 4, 5])

result = arr * 2

print(result)
```

```python
# 2. Broadcasting

a = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

b = np.array([10, 20, 30])

print(a + b)
```

```python
#vector normalization

arr=np.array(
[
    [1,2],[3,4]
])

mean=arr.mean();
std_dev=arr.std();

print(std_dev);

v_nor=(arr-mean)/std_dev;
print(v_nor);
```
