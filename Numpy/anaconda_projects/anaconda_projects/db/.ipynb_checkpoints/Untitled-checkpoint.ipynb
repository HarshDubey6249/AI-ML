```python
import numpy as np
```

```python
arr=np.array([1,2,3,4,5]);
print(arr)
```

```python
# 0D array
arr0d=np.array(23);
#1d array

arr1d=np.array([1,2,3,4,5]);
# 2d array

arr2d=np.array([[1,2],[3,4],[5,6]]);
# 3d array

arr3d=np.array([[[1,2,3],[11,12,13]],[[21,22,23,],[31,32,33]]]);
print(" 0D array");
print(arr0d)


print(" 1D array");
print(arr1d)


print(" 2D array");
print(arr2d)

print(" 0D array");
print(arr3d)
```

```python

a = np.array(42)
b = np.array([1, 2, 3, 4, 5])
c = np.array([[1, 2, 3], [4, 5, 6]])
d = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```
